# Titanic Dataset - Practice Challenge

```python
import pandas as pd
import numpy as np
titanic = pd.read_csv('./titanic_dataset.csv')
titanic_updated = titanic.copy()
```

```python
titanic_updated['Embarked'] = titanic_updated['Embarked'].fillna(titanic_updated['Embarked'].mode()[0])
```

```python
titanic_updated['Age'] = titanic_updated['Age'].fillna(titanic_updated['Age'].median())
```

```python
titanic_updated[titanic_updated['Cabin'].isna()][['Name', 'Age', 'Ticket', 'Fare', 'Cabin']]
```

```python
titanic_updated['hasCabin'] = titanic_updated['Cabin'].apply(lambda x: 0 if pd.isna(x) else 1)
```

```python
titanic_updated = titanic_updated.drop('Cabin', axis=1)
```

```python
titanic_updated = titanic_updated.drop(['PassengerId', 'Ticket'], axis=1)
```

```python
titanic_updated.info()
```

```python
titanic_updated.to_csv('./updated_titanic_dataset.csv', index=False)
```

```python
peopleSurvived = titanic_updated['Survived'].sum()
overallSurvivalRate = (peopleSurvived / 891) * 100
print(overallSurvivalRate)
```
