```python
import pandas as pd
import numpy as np

hp_data = pd.read_csv('./house_price_data.csv')
hp_updated = hp_data.copy()
```

```python
hp_updated = hp_updated.drop(['Unnamed: 0'], axis=1)
hp_updated
```

```python
hp_updated = hp_updated.rename({'Area in sq ft':'Area (sq ft)', 'No. of Bedrooms': 'Bedrooms', 'No. of Bathrooms': 'Bathrooms', 'No. of Receptions': 'Receptions'}, axis=1)
```

```python
hp_updated['Location'] = hp_updated['Location'].fillna("Unknown")
```

```python
hp_updated.info()
```

```python
hp_updated['Price per sqft'] = (hp_updated['Price'] / hp_updated['Area (sq ft)'])
```

```python
hp_updated
```

```python
hp_updated.to_csv('./updated_house_price_data.csv', index=False)
```

```python
hp_updated.groupby(['Location']).agg({'Area (sq ft)':'mean', 'Price':'mean'})
```

```python
hp_updated.groupby(['Bedrooms', 'Bathrooms', 'Receptions'])['Price'].median()
```

```python
hp_updated.groupby(['Bedrooms', 'Bathrooms', 'Receptions'])['Area (sq ft)'].median()
```

```python
hp_updated.groupby('Bedrooms')['Price per sqft'].median()
```

```python
hp_updated = hp_updated.drop(
    hp_updated[(hp_updated['Bathrooms'] == 0) & (hp_updated['Bedrooms'] == 0) & (hp_updated['Receptions'] == 0)].index
)
```

```python
hp_updated = hp_updated.drop(
    hp_updated[
        (hp_updated['Bathrooms'] == 8) &
        (hp_updated['Bedrooms'] == 8) &
        (hp_updated['Receptions'] == 8)
    ].index
)
hp_updated = hp_updated.drop(
    hp_updated[
        (hp_updated['Bathrooms'] == 9) &
        (hp_updated['Bedrooms'] == 9) &
        (hp_updated['Receptions'] == 9)
    ].index
)
hp_updated = hp_updated.drop(
    hp_updated[
        (hp_updated['Bathrooms'] == 10) &
        (hp_updated['Bedrooms'] == 10) &
        (hp_updated['Receptions'] == 10)
    ].index
)

hp_updated['Bedrooms'].value_counts()
```
