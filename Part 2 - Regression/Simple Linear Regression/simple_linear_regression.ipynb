# Simple Linear Regression

## Data Preprocessing

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('./Salary_Data.csv')

X = np.array(df.iloc[:, :-1])
y = np.array(df.iloc[:, -1])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

## Training the Simple Linear Regression model on the Training set

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

## Predicting Test set Results

```python
y_predicted = regressor.predict(X_test)
y_predicted
```

## Visualizing the Training Set Results

```python
# Real Salaries of the Training set
plt.scatter(X_train, y_train, color='red')

# Predicted Salaries of the Training set
plt.plot(X_train, regressor.predict(X_train), color='blue')

plt.title("Salary v/s Experience (Training Set)")
plt.xlabel("Experience (in years)")
plt.ylabel("Salary")

plt.show()
```

## Visualizing the Test set Results

```python
# Real Salaries of the Test set
plt.scatter(X_test, y_test, color='red')

# Predicted Salaries of the Test set
plt.plot(X_test, y_predicted, color='blue')

plt.title("Salary v/s Experience (Test Set)")
plt.xlabel("Experience (in years)")
plt.ylabel("Salary")

plt.show()
```

## Making predictions for new observations 

```python
# Predict the salary of an employee with 15 years of experience
print(regressor.predict([[15]]))

# Predict the salary of an employee with 14.2 years of experience
print(regressor.predict([[14.2]]))
```

## Getting the Regression coefficients

```python
slope = regressor.coef_[0]
intercept = regressor.intercept_

print(f"Slope = {slope}")
print(f"Intercept = {intercept}")
print("\nThe equation of the regression line is :-")
print(f"y = {slope}x + {intercept}")
```
