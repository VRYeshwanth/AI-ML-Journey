```python
import numpy as np
import pandas as pd

df = pd.read_csv('./data.csv')

X = df.iloc[:, :-1]
y = df.iloc[:, -1]
print(X)
```

```python
def standardScaler(arr):
    data = np.array(arr)

    mean = np.mean(data, axis=0)
    std = np.std(data, axis=0)

    scaled_data = (data - mean) / std
    return scaled_data, mean, std

X, X_mean, X_std = standardScaler(X)
y, y_mean, y_std = standardScaler(y)
```

```python
m = 0
c = 0
n = len(X)

alpha = 0.01 # Learning Rate
epochs = 1000
```

```python
for i in range(epochs):
    m_gradient = 0
    c_gradient = 0
    cost = 0

    for j in range(n):
        y_predicted = m * X[j] + c
        error = y[j] - y_predicted
        cost += (y[j] - y_predicted) ** 2
        m_gradient += (X[j] * error)
        c_gradient += error
    
    m_gradient = m_gradient * (-2 / n)
    c_gradient = c_gradient * (-2 / n)
    cost /= n

    m -= (alpha * m_gradient)
    c -= (alpha * c_gradient)

    if i % 50 == 0:
        print(f"Epoch {i} : Cost : {cost}")
```

```python
print("Scaled values are :-")
print(f"Slope (m) = {m}, Intercept (c) = {c}")

m_actual = (m * y_std) / X_std
c_actual = y_mean + (c * y_std) - ((X_mean * m * y_std) / X_std)

print("\nActual Values are :-")
print(f"Slope (m) = {m_actual}, Intercept (c) = {c_actual}")
```
