```python
nums = [1,2,3,4,5,6]
print("Original list :- ")
print(nums)

nums.append(7)
nums.append(8)
print("List after appending :- ")
print(nums)

nums.pop()
print("List after popping once :- ")
print(nums)

print("nums[0] = " +str(nums[0]))
print("nums[1:3] = " +str(nums[1:3]))

squares = [x*x for x in nums]
print("Squares of nums :- ")
print(squares)
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print("Hi, I am", self.name)
        print(f"I am {self.age} years old")

s1 = Student("Tester", 20)
s1.greet()
```
