# Data Preprocessing Tools

## Importing the libraries required

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Importing the dataset

```python
df = pd.read_csv('./Data.csv')
```

```python
# Features Matrix of the dataset
# Contains all the features of the dataset
X = df.iloc[:, :-1].values

# Dependant variable vector
y = df.iloc[:, -1].values
```

```python
print(X)
```

```python
print(y)
```

## Taking care of missing data

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer.fit(X[:, 1:3]) # Looks for all missing values in the columns age and salary
X[:, 1:3] = imputer.transform(X[:, 1:3]) # Updates the missign values with the mean value of the column and assigns it
```

```python
print(X)
```

## Encoding Categorical Data

### Encoding Independent Variable

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Transformers option requires 3 parameters :
# Type of transformation to be performed = 'encoder'
# Class which performs the above transformation = OneHotEncoder()
# Indexes of the columns to perform the transformations = 0

# remainder parameter 'passthrough' ensures that the other columns for which OneHotEncoding() is not applied will also be included. To not include them, we can exclude this parameter

# One Hot Encoding is used to encode multiple categories of given columns in a dataset
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [0])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

```python
print(X)
```

### Encoding dependant variable

```python
from sklearn.preprocessing import LabelEncoder

# Label Encoding is used when we have only two cateories in a column
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
print(y)
```

## Splitting Data into Training Set and Test Set

```python
from sklearn.model_selection import train_test_split

# test_train_split function splits the dataset into 4 parts :
# X_train = Contains the features matrix of the training set
# X_test = Contains the dependant variable vector of the training set
# y_train = Contains the features matrix of the test set
# y_test = Contains the dependant variable vector of the test set

# test_size parameter decides the percent of entries to be present in the test set i.e. 0.2 implies 20%
# random_state = 1, ensures that the split is not randomly changes every time the line executes
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
print(X_train)
```

```python
print(X_test)
```

```python
print(y_train)
```

```python
print(y_test)
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

# The line below calculates the mean and standard deviation for the numeric columns and then transforms it in the training feature set
X_train[:, 3:] = sc.fit_transform(X_train[:, 3:])

# The line below uses the same scalar as the training set to just transform the test feature set. We should not fit a new scalar for the test set.
X_test[:, 3:] = sc.transform(X_test[:, 3:])
```
