# Mini Project: 3-Feature Linear Regression

## Dataset

Generate a matrix:

$$
X \in \mathbb{R}^{200 \times 3}
$$

Use a normal distribution.

---

## Target Variable

Let:

$$
x_1 = X[:, 0]
$$

$$
x_2 = X[:, 1]
$$

$$
x_3 = X[:, 2]
$$

Construct:

$$
y = 6x_1 - 4x_2 + 2x_3 + 10
$$

---

## Design Matrix

Construct:

$$
X_b =
\begin{bmatrix}
1 & x_1 & x_2 & x_3
\end{bmatrix}
$$

So that:

$$
X_b \in \mathbb{R}^{200 \times 4}
$$

---

## Parameter Estimation (Normal Equation)

Compute:

$$
\theta = (X_b^T X_b)^{-1} X_b^T y
$$

---

## Prediction

$$
\hat{y} = X_b \theta
$$

---

## Mean Squared Error

$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

Where:

$$
n = 200
$$

---

## Extra Challenge (Harder)

Modify the target equation to:

$$
y = 6x_1 - 4x_2 + 2x_3 + 3x_1x_2 - 5x_2x_3 + 10
$$

Expand the feature matrix accordingly and solve again.

```python
import numpy as np

x = np.random.randn(200,3)
x12 = x[:,0] * x[:,1]
x23 = x[:,1] * x[:,2]
y = 6*x[:,0] - 4*x[:,1] + 2*x[:,2] + 3*(x12) - 5*(x23) + 10

ones = np.ones((200, 1))
x_b = np.hstack((ones, x, np.reshape(x12,(200,1)), np.reshape(x23,(200,1))))

theta = np.linalg.inv(x_b.T @ x_b) @ (x_b.T @ y)
print(theta)
```
