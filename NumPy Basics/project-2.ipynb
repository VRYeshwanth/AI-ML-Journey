# Mini Project: Single Feature Polynomial Regression

## Dataset

Generate:

$$
X \in \mathbb{R}^{150 \times 1}
$$

Use a normal distribution.

---

## Target Variable

Let:

$$
x = X[:, 0]
$$

Construct:

$$
y = 4x^2 - 3x + 5
$$

---

## Feature Expansion

Construct the design matrix:

$$
X_b =
\begin{bmatrix}
1 & x & x^2
\end{bmatrix}
$$

So that:

$$
X_b \in \mathbb{R}^{150 \times 3}
$$

---

## Parameter Estimation

Compute:

$$
\theta = (X_b^T X_b)^{-1} X_b^T y
$$

---

## Prediction

$$
\hat{y} = X_b \theta
$$

---

## Mean Squared Error

$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

Where:

$$
n = 150
$$

```python
import numpy as np

x = np.random.randn(150,1)

x2 = x[:,0] * x[:,0]
y = 4 * x2 - 3 * x[:,0] + 5

ones = np.ones((150, 1))
x_b = np.hstack((ones, x, np.reshape(x2, (150, 1))))

theta = np.linalg.inv(x_b.T @ x_b) @ x_b.T @ y
print(theta)
```
