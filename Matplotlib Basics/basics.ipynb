# Matplotlib basics

## 1. Basic Plot using plot()

```python
import matplotlib.pyplot as plt

x = [2020,2021,2022,2023,2024]
y = [50,27,72,41,64]

# Abbreviations
# ms = markersize (Sets the size of the marker provided)
# mfc = markerfacecolor (Sets the color of the face of the marker)
# mec = markeredgecolor (Sets the edge color of the marker)


plt.plot(x, y,
         marker=".",
         ms=10,
         mfc="red",
         mec="red",
         linewidth=2
         )

# Sets the title of the plot
plt.title("Population", fontsize=20, family="Arial", fontweight="bold", color="blue")

# Sets the X and Y Labels for the respective axes
plt.xlabel("Year", fontsize=15, family="Arial", fontweight="bold")
plt.ylabel("No of people", fontsize=15, family="Arial", fontweight="bold")

# Force Matplotlib to display only the specified x values as tick marks on the x-axis
plt.xticks(x)

# Sets grid lines for better visualization of data
plt.grid()

plt.show()
```

## 2. Bar Chart

```python
categories = ["Fruits", "Vegetables", "Proteins", "Dairy", "Sweets", "Grains"]
values = [5,2,8,4,6,7]

# Creates a bar chart
plt.bar(categories, values)

plt.title("Food Consumption")

plt.xlabel("Food")
plt.ylabel("Quantity")

plt.show()
```

```python
categories = ["Fruits", "Vegetables", "Proteins", "Dairy", "Sweets", "Grains"]
values = [5,2,8,4,6,7]

# Creates a horizontal bar chart
plt.barh(categories, values)

plt.title("Food Consumption")

plt.xlabel("Quantity")
plt.ylabel("Food")

plt.show()
```

## 3. Pie Chart

```python
categories = ["Year 1", "Year 2", "Year 3", "Year 4"]
values = [250, 200, 150, 100]

colors = ["red", "yellow", "blue", "green"]

# Parameters
# labels = Sets the labels for all the pies
# autopct = Can be used to display the percentage of each pie. Uses % as a format specifier
# colors = Sets specific colors for the pies
# explode = Displaces specific pies outward depending on the value
# startangle = Rotates the pie chart by a specific angle

plt.pie(values,
        labels=categories,
        autopct="%1.2f%%",
        colors=colors,
        explode=[0.1,0,0,0],
        startangle=270
        )

plt.title("College")

plt.plot()

plt.show()
```
