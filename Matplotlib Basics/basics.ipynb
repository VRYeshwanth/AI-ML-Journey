# Matplotlib basics

## 1. Basic Plot using plot()

```python
import matplotlib.pyplot as plt

x = [2020,2021,2022,2023,2024]
y = [50,27,72,41,64]

# Abbreviations
# ms = markersize (Sets the size of the marker provided)
# mfc = markerfacecolor (Sets the color of the face of the marker)
# mec = markeredgecolor (Sets the edge color of the marker)


plt.plot(x, y,
         marker=".",
         ms=10,
         mfc="red",
         mec="red",
         linewidth=2
         )

# Sets the title of the plot
plt.title("Population", fontsize=20, family="Arial", fontweight="bold", color="blue")

# Sets the X and Y Labels for the respective axes
plt.xlabel("Year", fontsize=15, family="Arial", fontweight="bold")
plt.ylabel("No of people", fontsize=15, family="Arial", fontweight="bold")

# Force Matplotlib to display only the specified x values as tick marks on the x-axis
plt.xticks(x)

# Sets grid lines for better visualization of data
plt.grid()

plt.show()
```

## 2. Bar Chart

```python
categories = ["Fruits", "Vegetables", "Proteins", "Dairy", "Sweets", "Grains"]
values = [5,2,8,4,6,7]

# Creates a bar chart
plt.bar(categories, values)

plt.title("Food Consumption")

plt.xlabel("Food")
plt.ylabel("Quantity")

plt.show()
```

```python
categories = ["Fruits", "Vegetables", "Proteins", "Dairy", "Sweets", "Grains"]
values = [5,2,8,4,6,7]

# Creates a horizontal bar chart
plt.barh(categories, values)

plt.title("Food Consumption")

plt.xlabel("Quantity")
plt.ylabel("Food")

plt.show()
```

## 3. Pie Chart

```python
categories = ["Year 1", "Year 2", "Year 3", "Year 4"]
values = [250, 200, 150, 100]

colors = ["red", "yellow", "blue", "green"]

# Parameters
# labels = Sets the labels for all the pies
# autopct = Can be used to display the percentage of each pie. Uses % as a format specifier
# colors = Sets specific colors for the pies
# explode = Displaces specific pies outward depending on the value
# startangle = Rotates the pie chart by a specific angle

plt.pie(values,
        labels=categories,
        autopct="%1.2f%%",
        colors=colors,
        explode=[0.1,0,0,0],
        startangle=270
        )

plt.title("College")

plt.plot()

plt.show()
```

## 4. Scatter Plot

```python
# No of hours studied
x1 = [0,1,1,2,2,3,4,5,5,6,7,8,9,9]
# Grades obtained
y1 = [60,63,66,67,75,73,74,79,84,85,89,93,92,95]

# No of hours studied
x2 = [0,1,2,3,3,4,4,5,6,6,7,8,8,9]
# Grades obtained
y2 = [55,60,58,66,68,70,73,75,78,80,83,84,88,90]

# Scatter Plots are used to find correlations

plt.scatter(x1, y1,
            color="red",
            alpha=0.5,
            s=50,
            label="Class A"
            )

plt.scatter(x2, y2,
            color="green",
            alpha=0.5,
            s=50,
            label="Class B"
            )

plt.title("Test Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Grades")

plt.legend()

plt.show()
```

## 5. Histograms

```python
import numpy as np

# Generates a normal distribution of numbers of size 100
# loc = Specifies the mean of the data i.e. all of the data points cluster around this value
# scale = Specifies the standard deviation of the data points. Higher the value, higher the deviation
scores = np.random.normal(loc=80, scale=10, size=100)

# Parameters
# bins = Specify the number of bins in the histogram
# color = Specify the color of the bars in the histogram
# edgecolor = Specify the edge color for the histogram (for better visibility)
plt.hist(scores,
         bins=10,
         color="green",
         edgecolor="black"
         )

plt.title("Exam Scores")
plt.xlabel("Score")
plt.ylabel("No of Students")

plt.show()
```

## 5. Subplots

```python
# The subplots() function is used to create multiple plots in a single entity called the Frame

# The subplot() function returns a tuple with two entities :-
# 1) Figure object - It is the entire canvas
# 2) A 2D matrix of axis objects - Contains all the axes where the plots will be plotted
figure, axes = plt.subplots(2,2)

x = np.array([1,1.5,2,2.5,3,3.5,4,4.5,5])

axes[0,0].plot(x, x*2, color="red")
axes[0,0].set_title("2x")

axes[0,1].plot(x, x**2, color="blue")
axes[0,1].set_title("x^2")

axes[1,0].plot(x, x**4, color="green")
axes[1,0].set_title("x^4")

axes[1,1].plot(x, x**6, color="brown")
axes[1,1].set_title("x^6")

# Provides spacing between the plots
plt.tight_layout()

plt.show()
```
