```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('./updated_house_price_data.csv')
```

```python
data.info()
```

```python
house_price = data.groupby('House Type')['Price'].median()

plt.figure(figsize=(12, 8))

plt.title("Median House Pricings", fontweight="bold", fontsize=20)
plt.xlabel("House Type", fontweight="bold", fontsize=20)
plt.ylabel("Price (in Million)", fontweight="bold", fontsize=20)

plt.bar(house_price.index, house_price.values / 1e6, edgecolor="black")

plt.ticklabel_format(style='sci', axis='y')
plt.grid(axis='y')
plt.tight_layout()

plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

plt.show()
```

```python
plt.figure(figsize=(12,8))

plt.title("Distribution of Price per Sq Ft", fontweight="bold", fontsize=20)
plt.xlabel("Price per Sq Ft", fontweight="bold", fontsize=20)
plt.ylabel("Frequency", fontweight="bold", fontsize=20)

plt.hist(data['Price per sqft'], bins=100, edgecolor="black")
plt.tight_layout()

plt.show()
```

```python
plt.title("Area vs Price", fontweight="bold", fontsize=20)
plt.xlabel("Area", fontweight="bold", fontsize=20)
plt.ylabel("Price (in Millions)", fontweight="bold", fontsize=20)

plt.scatter(data['Area (sq ft)'], data['Price'] / 1e6, alpha=0.3)
plt.xscale('log')
plt.yscale('log')

plt.tight_layout()

plt.show()
```

```python
plt.title("Area vs Price per sq.ft", fontweight="bold", fontsize=20)
plt.xlabel("Area", fontweight="bold", fontsize=20)
plt.ylabel("Price per sq.ft (in Millions)", fontweight="bold", fontsize=20)

plt.scatter(data['Area (sq ft)'], data['Price per sqft'] / 1e6, alpha=0.3)
plt.xscale('log')
plt.yscale('log')

plt.tight_layout()

plt.show()
```
